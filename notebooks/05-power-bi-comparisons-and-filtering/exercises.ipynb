## About the data

`amazon_monthly.csv` aggregates the same Amazon extract by month, shipping state and product category. Each row contains sample spending in USD, items purchased and the number of purchase records. There is no unique order identifier, so record counts are not order counts. Missing shipping states remain as Unknown. Months with no records for a particular state/category combination have no row.

Sources, licences, selection rules and limitations are in the [data documentation](../../data/README.md). Files are prepared by `scripts/fetch_data.py`; this notebook reads local data only.
